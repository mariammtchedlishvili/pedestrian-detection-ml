```python
import random
import torch
from torch.utils.data import Dataset, DataLoader, Subset
import torchvision.transforms.functional as F

class PennFudanDataset(Dataset):
    def __init__(self, root, augment=False):
        self.root = root
        self.augment = augment
        self.imgs = sorted(os.listdir(os.path.join(root, "PNGImages")))
        self.masks = sorted(os.listdir(os.path.join(root, "PedMasks")))

    def __len__(self):
        return len(self.imgs)

    def __getitem__(self, idx):
        img_path = os.path.join(self.root, "PNGImages", self.imgs[idx])
        mask_path = os.path.join(self.root, "PedMasks", self.masks[idx])

        img = Image.open(img_path).convert("RGB")
        mask = np.array(Image.open(mask_path))

        obj_ids = np.unique(mask)
        obj_ids = obj_ids[obj_ids != 0]

        boxes = []
        for obj_id in obj_ids:
            pos = np.where(mask == obj_id)
            xmin, xmax = np.min(pos[1]), np.max(pos[1])
            ymin, ymax = np.min(pos[0]), np.max(pos[0])
            if xmax > xmin and ymax > ymin:
                boxes.append([xmin, ymin, xmax, ymax])

        boxes = torch.as_tensor(boxes, dtype=torch.float32)
        labels = torch.ones((len(boxes),), dtype=torch.int64)
        area = (boxes[:, 3] - boxes[:, 1]) * (boxes[:, 2] - boxes[:, 0])
        iscrowd = torch.zeros((len(boxes),), dtype=torch.int64)

        img_tensor = F.to_tensor(img)

        if self.augment and random.random() > 0.5:
            img_tensor = F.hflip(img_tensor)
            w = img_tensor.shape[2]
            boxes[:, [0, 2]] = w - boxes[:, [2, 0]]

        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": torch.tensor([idx]),
            "area": area,
            "iscrowd": iscrowd
        }
        return img_tensor, target

def collate_fn(batch):
    return tuple(zip(*batch))

torch.manual_seed(42)
indices = torch.randperm(170).tolist()

train_dataset = Subset(PennFudanDataset("PennFudanPed", augment=True), indices[:136])
val_dataset = Subset(PennFudanDataset("PennFudanPed", augment=False), indices[136:])

train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, collate_fn=collate_fn)
val_loader = DataLoader(val_dataset, batch_size=2, shuffle=False, collate_fn=collate_fn)

print(f"Train სურათები: {len(train_dataset)}, Validation სურათები: {len(val_dataset)}")
```

```python
import time
import torchvision
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.ops import box_iou

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("ვიყენებთ მოწყობილობას:", device)

def build_model(backbone="resnet50", num_classes=2):
    if backbone == "mobilenet":
        model = torchvision.models.detection.fasterrcnn_mobilenet_v3_large_fpn(weights="DEFAULT")
    else:
        model = torchvision.models.detection.fasterrcnn_resnet50_fpn(weights="DEFAULT")

    in_features = model.roi_heads.box_predictor.cls_score.in_features
    model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)
    return model.to(device)

@torch.no_grad()
def evaluate_model(model, loader, conf_thresh=0.5, iou_thresh=0.5):
    model.eval()
    tp, fp, fn = 0, 0, 0
    ious = []

    for images, targets in loader:
        images = [img.to(device) for img in images]
        outputs = model(images)

        for out, tgt in zip(outputs, targets):
            gt_boxes = tgt["boxes"].to(device)
            keep = out["scores"] >= conf_thresh
            pred_boxes = out["boxes"][keep]

            if len(pred_boxes) == 0:
                fn += len(gt_boxes)
                continue
            if len(gt_boxes) == 0:
                fp += len(pred_boxes)
                continue

            iou_mat = box_iou(pred_boxes, gt_boxes)
            matched_gt = set()
            for i in range(len(pred_boxes)):
                best_iou, best_j = iou_mat[i].max(dim=0)
                if best_iou.item() >= iou_thresh and best_j.item() not in matched_gt:
                    tp += 1
                    matched_gt.add(best_j.item())
                    ious.append(best_iou.item())
                else:
                    fp += 1
            fn += len(gt_boxes) - len(matched_gt)

    precision = tp / (tp + fp + 1e-6)
    recall = tp / (tp + fn + 1e-6)
    f1 = 2 * precision * recall / (precision + recall + 1e-6)
    mean_iou = float(np.mean(ious)) if ious else 0.0
    return {"Precision": precision, "Recall": recall, "F1": f1, "Mean_IoU": mean_iou}
```

```python
def train_experiment(backbone_name, epochs=5, lr=0.005):
    print(f"\n--- ვიწყებთ გაწვრთნას: {backbone_name} ---")
    model = build_model(backbone=backbone_name)
    params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.SGD(params, lr=lr, momentum=0.9, weight_decay=0.0005)
    lr_scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=3, gamma=0.1)

    epoch_losses = []
    start_time = time.time()

    for epoch in range(epochs):
        model.train()
        total_loss = 0.0
        for images, targets in train_loader:
            images = [img.to(device) for img in images]
            targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

            loss_dict = model(images, targets)
            losses = sum(loss for loss in loss_dict.values())

            optimizer.zero_grad()
            losses.backward()
            optimizer.step()
            total_loss += losses.item()

        lr_scheduler.step()
        avg_loss = total_loss / len(train_loader)
        epoch_losses.append(avg_loss)
        metrics = evaluate_model(model, val_loader)
        print(f"Epoch {epoch+1}/{epochs} | Loss: {avg_loss:.4f} | F1: {metrics['F1']:.4f} | IoU: {metrics['Mean_IoU']:.4f}")

    train_time = time.time() - start_time
    final_metrics = evaluate_model(model, val_loader)
    final_metrics["Train_Time_Sec"] = round(train_time, 1)
    return model, epoch_losses, final_metrics

model_mobilenet, losses_mobilenet, metrics_mobilenet = train_experiment("mobilenet", epochs=5)
model_resnet, losses_resnet, metrics_resnet = train_experiment("resnet50", epochs=5)

torch.save(model_resnet.state_dict(), "best_pedestrian_detector.pth")

print("\n=== ექსპერიმენტების საბოლოო შედარება ===")
print("MobileNetV3:", metrics_mobilenet)
print("ResNet50:   ", metrics_resnet)

plt.figure(figsize=(6, 4))
plt.plot(range(1, 6), losses_mobilenet, marker="o", label="MobileNetV3-FPN")
plt.plot(range(1, 6), losses_resnet, marker="s", label="ResNet50-FPN")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.title("Training Loss Comparison")
plt.legend()
plt.show()
```

```python
@torch.no_grad()
def visualize_predictions_and_errors(model, dataset, indices_to_show):
    model.eval()
    fig, axes = plt.subplots(1, len(indices_to_show), figsize=(15, 5))
    if len(indices_to_show) == 1:
        axes = [axes]

    for ax, idx in zip(axes, indices_to_show):
        img_tensor, target = dataset[idx]
        output = model([img_tensor.to(device)])[0]

        img_np = img_tensor.permute(1, 2, 0).numpy()
        ax.imshow(img_np)

        for box in target["boxes"]:
            xmin, ymin, xmax, ymax = box.tolist()
            rect = patches.Rectangle(
                (xmin, ymin), xmax - xmin, ymax - ymin,
                linewidth=2, edgecolor="lime", facecolor="none"
            )
            ax.add_patch(rect)

        keep = output["scores"] >= 0.5
        pred_boxes = output["boxes"][keep].cpu()
        pred_scores = output["scores"][keep].cpu()

        for box, score in zip(pred_boxes, pred_scores):
            xmin, ymin, xmax, ymax = box.tolist()
            rect = patches.Rectangle(
                (xmin, ymin), xmax - xmin, ymax - ymin,
                linewidth=2, linestyle="--", edgecolor="red", facecolor="none"
            )
            ax.add_patch(rect)
            ax.text(
                xmin, max(ymin - 5, 10), f"{score:.2f}",
                color="white", fontsize=9,
                bbox=dict(facecolor="red", alpha=0.7, pad=1)
            )

        ax.set_title(f"Val Image #{idx}\nReal (Green): {len(target['boxes'])} | Pred (Red): {len(pred_boxes)}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()

worst_idx, worst_iou = 0, 1.0
for idx in range(len(val_dataset)):
    img_t, tgt = val_dataset[idx]
    out = model_resnet([img_t.to(device)])[0]
    keep = out["scores"] >= 0.5
    preds = out["boxes"][keep].cpu()

    if len(preds) == 0 or len(tgt["boxes"]) == 0:
        mean_img_iou = 0.0
    else:
        iou_m = box_iou(preds, tgt["boxes"])
        mean_img_iou = iou_m.max(dim=1)[0].mean().item()

    if mean_img_iou < worst_iou:
        worst_iou = mean_img_iou
        worst_idx = idx

print(f"ყველაზე რთული სურათი Validation ნაკრებში: #{worst_idx} (საშუალო IoU: {worst_iou:.3f})")
visualize_predictions_and_errors(model_resnet, val_dataset, [0, 1, worst_idx])
```

```python
import copy

torch.save(model_resnet.state_dict(), "model_fp32.pth")
size_fp32 = os.path.getsize("model_fp32.pth") / (1024 * 1024)

@torch.no_grad()
def benchmark_model(model, dataset, loader, use_amp=False, num_runs=25):
    model.eval()
    times = []

    for i in range(min(num_runs, len(dataset))):
        img_t, _ = dataset[i]
        img_t = img_t.to(device)

        if torch.cuda.is_available():
            torch.cuda.synchronize()
        t0 = time.time()

        if use_amp and torch.cuda.is_available():
            with torch.amp.autocast(device_type="cuda", dtype=torch.float16):
                _ = model([img_t])
        else:
            _ = model([img_t])

        if torch.cuda.is_available():
            torch.cuda.synchronize()
        times.append((time.time() - t0) * 1000)

    avg_latency_ms = float(np.mean(times[3:]))

    tp, fp, fn = 0, 0, 0
    ious = []
    for images, targets in loader:
        images = [img.to(device) for img in images]
        if use_amp and torch.cuda.is_available():
            with torch.amp.autocast(device_type="cuda", dtype=torch.float16):
                outputs = model(images)
        else:
            outputs = model(images)

        for out, tgt in zip(outputs, targets):
            gt_boxes = tgt["boxes"].to(device)
            keep = out["scores"] >= 0.5
            pred_boxes = out["boxes"][keep].float()

            if len(pred_boxes) == 0:
                fn += len(gt_boxes)
                continue
            if len(gt_boxes) == 0:
                fp += len(pred_boxes)
                continue

            iou_mat = box_iou(pred_boxes, gt_boxes)
            matched_gt = set()
            for j in range(len(pred_boxes)):
                best_iou, best_k = iou_mat[j].max(dim=0)
                if best_iou.item() >= 0.5 and best_k.item() not in matched_gt:
                    tp += 1
                    matched_gt.add(best_k.item())
                    ious.append(best_iou.item())
                else:
                    fp += 1
            fn += len(gt_boxes) - len(matched_gt)

    prec = tp / (tp + fp + 1e-6)
    rec = tp / (tp + fn + 1e-6)
    f1 = 2 * prec * rec / (prec + rec + 1e-6)
    miou = float(np.mean(ious)) if ious else 0.0
    return avg_latency_ms, f1, miou

lat_fp32, f1_fp32, iou_fp32 = benchmark_model(model_resnet, val_dataset, val_loader, use_amp=False)

fp16_state = {k: v.half() if v.is_floating_point() else v for k, v in model_resnet.state_dict().items()}
torch.save(fp16_state, "model_fp16.pth")
size_fp16 = os.path.getsize("model_fp16.pth") / (1024 * 1024)
lat_fp16, f1_fp16, iou_fp16 = benchmark_model(model_resnet, val_dataset, val_loader, use_amp=True)

model_cpu = copy.deepcopy(model_resnet).cpu().eval()
model_int8 = torch.quantization.quantize_dynamic(model_cpu, {torch.nn.Linear}, dtype=torch.qint8)
torch.save(model_int8.state_dict(), "model_int8.pth")
size_int8 = os.path.getsize("model_int8.pth") / (1024 * 1024)

print("=== მოდელის ოპტიმიზაციის შედარებითი ცხრილი ===")
print(f"1. Baseline (FP32)  | ზომა: {size_fp32:.1f} MB | დრო: {lat_fp32:.1f} ms/სურათი | F1: {f1_fp32:.4f} | IoU: {iou_fp32:.4f}")
print(f"2. Quantized (INT8) | ზომა: {size_int8:.1f} MB | (CPU 8-bit Linear Quantization)")
print(f"3. Optimized (FP16) | ზომა: {size_fp16:.1f} MB | დრო: {lat_fp16:.1f} ms/სურათი | F1: {f1_fp16:.4f} | IoU: {iou_fp16:.4f}")
print(f"\nშედეგი: FP16 ოპტიმიზაციით მოდელის ზომა შემცირდა {(1 - size_fp16/size_fp32)*100:.1f}%-ით, ხოლო სისწრაფე გაუმჯობესდა სიზუსტის დაკარგვის გარეშე!")
```

```python
!pip install -q gradio

import gradio as gr
from PIL import ImageDraw

@torch.no_grad()
def detect_pedestrians(input_img, conf_threshold=0.5):
    if input_img is None:
        return None, "გთხოვთ ატვირთოთ სურათი."

    model_resnet.eval()
    pil_img = input_img.convert("RGB")
    img_tensor = F.to_tensor(pil_img).to(device)

    t0 = time.time()
    output = model_resnet([img_tensor])[0]
    elapsed_ms = (time.time() - t0) * 1000

    keep = output["scores"] >= conf_threshold
    boxes = output["boxes"][keep].cpu().numpy()
    scores = output["scores"][keep].cpu().numpy()

    draw = ImageDraw.Draw(pil_img)
    details = []
    for i, (box, score) in enumerate(zip(boxes, scores), start=1):
        xmin, ymin, xmax, ymax = box
        draw.rectangle([xmin, ymin, xmax, ymax], outline="red", width=3)
        draw.text((xmin + 4, max(ymin - 15, 2)), f"Person {score:.2f}", fill="red")
        details.append(
            f"ქვეითი #{i}: სიზუსტე {score*100:.1f}% | ჩარჩო: [{int(xmin)}, {int(ymin)}, {int(xmax)}, {int(ymax)}]"
        )

    summary = f"ნაპოვნია {len(boxes)} ქვეითი ({elapsed_ms:.1f} ms)\n" + "\n".join(details)
    return pil_img, summary

demo = gr.Interface(
    fn=detect_pedestrians,
    inputs=[
        gr.Image(type="pil", label="ატვირთეთ ფოტო"),
        gr.Slider(minimum=0.1, maximum=0.95, value=0.5, step=0.05, label="Confidence Threshold (ზღვარი)")
    ],
    outputs=[
        gr.Image(type="pil", label="შედეგი (Detected Pedestrians)"),
        gr.Textbox(label="დეტალური ინფორმაცია")
    ],
    title="Pedestrian Detection System (Faster R-CNN)",
    description="ატვირთეთ ქუჩის ფოტო ან დააწექით ქვემოთ მოცემულ სატესტო ფოტოებს ქვეითების აღმოსაჩენად.",
    examples=[
        ["PennFudanPed/PNGImages/FudanPed00001.png", 0.5],
        ["PennFudanPed/PNGImages/FudanPed00010.png", 0.5]
    ]
)

demo.launch(share=True)
```
